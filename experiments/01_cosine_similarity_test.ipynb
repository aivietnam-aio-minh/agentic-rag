```python
from sentence_transformers import SentenceTransformer

# Nạp model BGE-M3 (hệ thống sẽ lấy từ cache ổ đĩa bạn vừa tải lúc nãy rất nhanh)
model = SentenceTransformer("BAAI/bge-m3", device="cuda")
```

```python
import numpy as np
def cosine(a,b):
    dot_product = np.dot(a,b)
    norm_a = np.linalg.norm(a)
    norm_b = np.linalg.norm(b)
    return dot_product / (norm_a * norm_b)
```

```python
v1 = model.encode("Học lập trình AI cần tư duy logic.")
v2 = model.encode("Để viết mã trí tuệ nhân tạo, bạn phải suy luận mạch lạc.")
v3 = model.encode("Trưa nay tôi thích ăn bún chả Hà Nội.")

print("Độ tương đồng giữa Câu 1 và Câu 2 (Cùng nghĩa):", cosine(v1, v2))
print("Độ tương đồng giữa Câu 1 và Câu 3 (Khác chủ đề):", cosine(v1, v3))
```
